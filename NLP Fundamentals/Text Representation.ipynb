```python
import numpy as np
from sklearn.preprocessing import OneHotEncoder
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
corpus = [
    "I love Python programming",
    "Python is easy to learn",
    "I love machine learning",
    "Machine learning is amazing"
]
```

## ONE-HOT ENCODING

```python
words = [["python"], ["learning"], ["machine"]]

encoder = OneHotEncoder(sparse_output=False)
one_hot = encoder.fit_transform(words)

print("Features:", encoder.get_feature_names_out())
print(one_hot)
```

## BAG OF WORDS

```python
bow = CountVectorizer()
X_bow = bow.fit_transform(corpus)

print("Features:", bow.get_feature_names_out())
print(X_bow.toarray())
```

## BIGRAM

```python
bigram = CountVectorizer(ngram_range=(2, 2))
X_bigram = bigram.fit_transform(corpus)

print("Features:", bigram.get_feature_names_out())
print(X_bigram.toarray())
```

## TRIGRAM

```python
trigram = CountVectorizer(ngram_range=(3, 3))
X_trigram = trigram.fit_transform(corpus)

print("Features:", trigram.get_feature_names_out())
print(X_trigram.toarray())
```

## TF-IDF

```python
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(corpus)

print("Features:", tfidf.get_feature_names_out())
print(np.round(X_tfidf.toarray(), 2))
```

## CUSTOM FEATURES

```python
text = "I really love Python programming!!!"

custom_features = {
    "word_count": len(text.split()),
    "character_count": len(text),
    "exclamation_count": text.count("!"),
    "contains_python": int("python" in text.lower()),
    "contains_love": int("love" in text.lower())
}

print(custom_features)
```
