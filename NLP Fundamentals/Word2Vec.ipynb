```python
from gensim.models import Word2Vec
import re
```

```python
sentences = [
    ["i", "love", "python"],
    ["i", "love", "machine", "learning"],
    ["python", "is", "easy"],
    ["machine", "learning", "is", "interesting"],
    ["deep", "learning", "is", "powerful"],
    ["python", "is", "powerful"],
    ["i", "love", "deep", "learning"]
]

print("SAMPLE CORPUS")
print(sentences)
```

## CBOW MODEL

```python
cbow_model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1,
    sg=0,
    seed=42,
    epochs=100
)

print("CBOW VECTOR FOR PYTHON")
print(cbow_model.wv["python"])

print("\nCBOW SIMILAR WORDS")
print(cbow_model.wv.most_similar("python", topn=3))
```

## SKIP-GRAM MODEL

```python
skipgram_model = Word2Vec(
    sentences=sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1,
    sg=1,
    seed=42,
    epochs=100
)

print("SKIP-GRAM VECTOR FOR PYTHON")
print(skipgram_model.wv["python"])

print("\nSKIP-GRAM SIMILAR WORDS")
print(skipgram_model.wv.most_similar("python", topn=3))
```

## WORD SIMILARITY

```python
similarity = skipgram_model.wv.similarity(
    "machine",
    "learning"
)

print("WORD SIMILARITY")
print(similarity)
```

## VOCABULARY

```python
print("VOCABULARY")
print(skipgram_model.wv.index_to_key)
```

## VECTOR SIZE

```python
print("VECTOR SIZE")
print(skipgram_model.wv["python"].shape)
```

## TRAIN WORD2VEC ON YOUR OWN TEXT

```python
text = """
Python is easy to learn.
I love Python programming.
Machine learning is interesting.
Deep learning is powerful.
Python helps with machine learning.
"""

text = text.lower()
text = re.sub(r"[^a-z\s]", "", text)

own_sentences = [
    line.split()
    for line in text.splitlines()
    if line.strip()
]

print("TOKENIZED OWN DATA")
print(own_sentences)

own_model = Word2Vec(
    sentences=own_sentences,
    vector_size=50,
    window=2,
    min_count=1,
    workers=1,
    sg=1,
    seed=42,
    epochs=100
)

print("\nOWN DATA - SIMILAR WORDS")
print(own_model.wv.most_similar("python", topn=3))

# Save the trained model
own_model.save("my_word2vec.model")
```
