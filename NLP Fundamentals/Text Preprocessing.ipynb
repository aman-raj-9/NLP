```python
import re
import string
from bs4 import BeautifulSoup
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
text = """
<HTML>
<P>The CATS are running quickly on the STREETS!</P>
<p>I am LOVING this amazing website: https://www.example.com</p>
<div>The dogs were barking, and the children are playing happily.</div>
<p>Visit www.google.com for MORE information!!!</p>
</HTML>
"""
```

```python
text = text.upper()
print(text)
```

```python
text = text.lower()
print(text)
```

## Remove HTML

```python
R_html = BeautifulSoup(text,"html.parser").get_text()
```

```python
print(R_html)
```

## Remove URLs

```python
R_urls = re.sub(r'https?://\S+|www\.\S+', '', R_html)
```

```python
print(R_urls)
```

## Remove Punctuation

```python
R_p = R_urls.translate(
    str.maketrans('', '', string.punctuation)
)
```

```python
print(R_p)
```

## Tokenization

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
tokens = word_tokenize(R_p)
```

```python
print(tokens)
```

## Remove stopwords

```python
stop_words = set(stopwords.words("english"))

S_t = [
    word for word in tokens
    if word not in stop_words
]
```

```python
print(S_t)
```

## Lemmatization

```python
lemmatizer = WordNetLemmatizer()

L = [
    lemmatizer.lemmatize(word)
    for word in S_t
]
```

```python
print(L)
```
