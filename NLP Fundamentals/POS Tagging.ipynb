```python
import nltk
nltk.download("punkt")
nltk.download("punkt_tab")
nltk.download("averaged_perceptron_tagger_eng")
```

```python
text = "The quick brown fox is running in the garden."
print(text)
```

## Tokenization

```python
tokens = nltk.word_tokenize(text)

print("Tokens:")
print(tokens)
```

## POS Tagging

```python
pos_tags = nltk.pos_tag(tokens)

print("POS Tags:")
for word, tag in pos_tags:
    print(word, "=", tag)
```
